## 🧮 Problem 1 Gradient Descent
## Gradient descent is a generic optimization algorithm capable of finding optimal solutions to a wide range of problems

We are given a loss function:

## 🧮 Problem 1 (15 pts) — Gradient Descent

Given the loss function:
**Given:**  

**Given:**

$$J(\theta_1, \theta_2) = \theta_1^2 + 4\theta_2^2 + \theta_1\theta_2$$

Initial parameters:

$$\theta_1^{(0)} = -3, \quad \theta_2^{(0)} = 3$$

Learning rate:

$$\alpha = 0.001$$


---

### Step 1 — Compute the Gradients


Find the partial derivatives of the loss function with respect to $\theta_1$ and $\theta_2$

$$
\frac{\partial J}{\partial \theta_1} = 2\theta_1 + \theta_2
$$

$$
\frac{\partial J}{\partial \theta_2} = 8\theta_2 + \theta_1
$$

Thus,

$$
\nabla J =
\begin{bmatrix}
2\theta_1 + \theta_2 \\
8\theta_2 + \theta_1
\end{bmatrix}
$$

---

### ### Step 2 — Evaluate at $(\theta_1, \theta_2) = (-3, 3)$

$$
\frac{\partial J}{\partial \theta_1} = 2(-3) + 3 = -3
$$

$$
\frac{\partial J}{\partial \theta_2} = 8(3) + (-3) = 21
$$

So,

$$
\nabla J =
\begin{bmatrix}
-3 \\
21
\end{bmatrix}
$$

---

### Step 3 — Gradient Descent Update

Rule:

$$
\theta^{(k+1)} = \theta^{(k)} - \alpha\nabla J
$$

Substitute values:

$$
\begin{bmatrix}
\theta_1^{(1)} \\
\theta_2^{(1)}
\end{bmatrix}
=
\begin{bmatrix}
-3 \\
3
\end{bmatrix}
-
0.001
\begin{bmatrix}
-3 \\
21
\end{bmatrix}
$$

Compute:

$$
\theta_1^{(1)} = -3 - 0.001(-3) = -2.997
$$

$$
\theta_2^{(1)} = 3 - 0.001(21) = 2.979
$$

---

### ✅ Final Solution

$$
(\theta_1,\theta_2) = (-2.997,\; 2.979)
$$

| Component | Expression | Value |
|:-----------|:-----------|:------|
| ∂J/∂θ₁ | \(2θ₁ + θ₂\) | −3 |
| ∂J/∂θ₂ | \(8θ₂ + θ₁\) | 21 |
| Updated θ₁ | \(−3 + 0.003\) | **−2.997** |
| Updated θ₂ | \(3 − 0.021\) | **2.979** |

**✅ Answer:** $((\theta_1, \theta_2) = (−2.997, 2.979)$)

### 🧮 Problem 2 — Polynomial Regression with L2 Regularization

Given 10 samples with two input variables \(x_1\) and \(x_2\).  
Fit a **polynomial regression of degree 3** using **L2 regularization** (Ridge Regression) to find the best-fitting equation.

---

#### Step 1 — Generate Data
The data are produced as:
$$[
y = 4x_1^3 + 3x_2^3 + 2x_1x_2 + \varepsilon
]
$$
where $(\varepsilon)$ is Gaussian noise.

---

#### Step 2 — Constant-One Column Trick
Using `PolynomialFeatures(degree=3, include_bias=True)` to automatically add a **column of ones** for the bias term.  
This implements the *constant-one column trick* in matrix form.

---

#### Step 3 — X Initialization for Least Squares
The polynomial expansion forms the design matrix $(X_{\text{poly}})$ that contains all monomials up to degree 3.

---

#### Step 4 — Least Squares Formulation with L2 Regularization
The Ridge regression solution minimizes:

$$
\min_{w}\, \|y - Xw\|^2 + \lambda\|w\|^2
$$

whose closed-form solution is:

$$
\hat{w} = (X^{T}X + \lambda I)^{-1} X^{T}y
$$
We implement this using `Ridge(alpha=λ)` in `scikit-learn`.

---


```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge

# --- Step 1: Generate data ---
np.random.seed(42)
X = np.random.rand(10, 2)       # shape (10, 2)
y = 4*X[:,0]**3 + 3*X[:,1]**3 + 2*X[:,0]*X[:,1] + np.random.randn(10)*0.1

# --- Step 2: Polynomial features with bias column ---
poly = PolynomialFeatures(degree=3, include_bias=True)
X_poly = poly.fit_transform(X)

# --- Step 3: Fit Ridge (L2) Regression ---
ridge = Ridge(alpha=0.1, fit_intercept=False)
ridge.fit(X_poly, y)

# --- Step 4: Display Equation ---
feature_names = poly.get_feature_names_out(['x1', 'x2'])
print("Polynomial Regression Equation (Degree 3 + L2 Regularization):\n")
for name, coef in zip(feature_names, ridge.coef_):
    print(f"{name:10s}: {coef:.4f}")
```

### 🧮 Problem 3 — Manual Implementation of L₁ Regularization (Lasso)

We apply **L₁ regularization** (Lasso) to a **degree-3 polynomial regression** on two variables.  
The goal is to understand how different regularization strengths \( \lambda \in \{0.1, 1, 10\} \) affect sparsity.

---

#### Step 1 — Objective Function

The loss function combines mean-squared error and the L₁ penalty:

$$
J(w,b) = \frac{1}{m}\sum_{i=1}^m (y^{(i)} - (\mathbf{x}^{(i)} \cdot w + b))^2
+ \lambda \sum_{j=1}^n |w_j|
$$

where  

- \(w\) = weight vector (coefficients excluding bias)  
- \(b\) = bias term  
- \(X\) = feature matrix  
- \(y\) = target vector  
- \(\lambda\) = regularization weight  

---

#### Step 2 — Gradient Descent Updates

For each iteration:

\[
\begin{aligned}
y_{\text{pred}} &= Xw + b \\
dw &= \frac{1}{m} X^T (y_{\text{pred}} - y) + \lambda\,\text{sign}(w) \\
db &= \frac{1}{m}\sum (y_{\text{pred}} - y)
\end{aligned}
\]

\[
\begin{aligned}
w &:= w - \alpha\,dw \\
b &:= b - \alpha\,db
\end{aligned}
\]

where  
\(\text{sign}(w_j)=1\) if \(w_j>0\), \(-1\) if \(w_j<0\), and \(0\) if \(w_j=0\).

---

#### Step 3 — Experimental Procedure

For each $(\lambda \in \{0.1, 1, 10\})$:

1. Loop over random seeds 1 to 10.  
2. Generate data using the given function.  
3. Apply gradient descent until convergence.  
4. Compute the ratio of zero coefficients (including bias).  
5. Compute the average ratio across 10 datasets.

---

#### Step 4 — Interpretation

Higher $(\lambda)$ increases penalty → more weights driven to zero → higher sparsity.


```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures

def lasso_gradient_descent(X, y, alpha=0.001, lambda_l1=0.1, iters=2000):
    """Manual gradient descent for L1-regularized regression."""
    m, n = X.shape
    w = np.zeros(n)
    b = 0.0

    for _ in range(iters):
        y_pred = X @ w + b
        dw = (1/m) * (X.T @ (y_pred - y)) + lambda_l1 * np.sign(w)
        db = (1/m) * np.sum(y_pred - y)

        w -= alpha * dw
        b -= alpha * db

    return w, b

def zero_ratio(w, b):
    """Ratio of zero coefficients including bias."""
    all_params = np.concatenate(([b], w))
    return np.sum(np.isclose(all_params, 0, atol=1e-4)) / len(all_params)

# Experiment parameters
weights = [0.1, 1, 10]
avg_zero_ratios = {}

for lam in weights:
    ratios = []
    for seed in range(1, 11):
        np.random.seed(seed)
        # Generate data
        X = np.random.rand(10, 2)
        y = 4*X[:,0]**3 + 3*X[:,1]**3 + 2*X[:,0]*X[:,1] + np.random.randn(10)*0.1

        # Polynomial features (degree 3)
        poly = PolynomialFeatures(degree=3, include_bias=True)
        X_poly = poly.fit_transform(X)

        # Train using manual L1 gradient descent
        w, b = lasso_gradient_descent(X_poly, y, alpha=0.001, lambda_l1=lam, iters=3000)
        ratios.append(zero_ratio(w, b))

    avg_zero_ratios[lam] = np.mean(ratios)

# Display results
print("Average ratio of zero coefficients (including bias):\n")
for lam, ratio in avg_zero_ratios.items():
    print(f"λ = {lam:<4}: {ratio:.2f}")
```
